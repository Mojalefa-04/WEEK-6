## Download AAPL

```python
import yfinance as yf

# Download approximately 5 years of Apple stock data
df = yf.download(
    "AAPL",
    start="2021-07-29",
    end="2026-07-30"
)

# Flatten multi-level columns
df.columns = df.columns.get_level_values(0)

# Save the dataset
df.to_csv("../Data/AAPL.csv")

# Display first 5 rows
df.head()
```
